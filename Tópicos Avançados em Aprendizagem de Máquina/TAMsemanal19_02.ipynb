# Pré-processamento

````python
from skimage.filters import gaussian
from skimage.filters import prewitt_h, prewitt_v
from skimage.morphology import disk
from skimage.morphology import binary_dilation, binary_erosion
from scipy import ndimage as ndi
from skimage.io import imread
import numpy as np

from skimage.measure import label, regionprops

from skimage.transform import resize
import matplotlib.pyplot as plt

# Realce dos atributos e detecção de bordas usando Prewitt
def normalize01(x):
    x = x.astype(np.float64)
    x = x - x.min()
    if x.max() > 0:
        x = x / x.max()
    return x

def grad_magnitude(gx, gy):
    return np.sqrt(gx**2 + gy**2)

def recorte_objeto(image):
    # Identifica as manchas brancas
    label_img = label(image)
    regions = regionprops(label_img)

    # Pega o maior objeto
    objeto_principal = max(regions, key=lambda r: r.area)
    minr, minc, maxr, maxc = objeto_principal.bbox

    # Faz o recorte automático
    return image[minr:maxr, minc:maxc]

def pre_processar(image):
    # Aplicação de filtro gaussiano
    resGaussian = gaussian(image, sigma=5)

    # Realce dos atributos e detecção de bordas usando Prewitt
    gx = prewitt_h(resGaussian)
    gy = prewitt_v(resGaussian)
    mag = grad_magnitude(gx, gy)

    normal_mag = normalize01(mag)

    # Aplicação de Limiar Thresholding para segmentação
    threshold_value = 0.2
    binary_segmentation = normal_mag > threshold_value

    # Usando um elemento estruturante menor para não apagar as bordas finas
    struc = disk(3)

    # Aplica a morfologia para fechar pequenos buracos no contorno
    dilated = binary_dilation(binary_segmentation, struc)
    closed = binary_erosion(dilated, struc)

    # identifica áreas pretas cercadas por áreas brancas e as preenche
    fill_object = ndi.binary_fill_holes(closed)

    # 1. Realiza o recorte do objeto
    return recorte_objeto(fill_object)


def pooling(image):
    image = image.astype(float)
    # Rendimensiona as imagens em tamanho fixo em 28x28
    return resize(image, (28,28), anti_aliasing=True)
````

# Evaluation Metric and Loss Function

````python
def accuracy(outputs, labels):
    _, preds = torch.max(outputs, dim = 1)
    return(torch.tensor(torch.sum(preds == labels).item()/ len(preds)))

print("Accuracy: ",accuracy(outputs, labels))
print("\n")
loss_fn = F.cross_entropy
print("Loss Function: ",loss_fn)
print("\n")
## Loss for the current batch
loss = loss_fn(outputs, labels)
print(loss)
````

**cross_entropy:** A métrica de erro. Ela diz ao modelo o quão longe a previsão está da realidade.

Diferente da perda, a acurácia é uma medida humana: qual a porcentagem de imagens que o modelo acertou? É o que usamos para decidir se o modelo está bom o suficiente para ser usado.

# Training the Model

````python
class MnistModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(input_size, num_classes)

    def forward(self, xb):
        # Adicionado os parênteses em .float()
        xb = xb.reshape(-1, 784).float()
        out = self.linear(xb)
        return(out)

    def training_step(self, batch):
        images, labels = batch
        out = self(images) ## Generate predictions
        loss = F.cross_entropy(out, labels) ## Calculate the loss
        return(loss)

    def validation_step(self, batch):
        images, labels = batch
        out = self(images)
        loss = F.cross_entropy(out, labels)
        acc = accuracy(out, labels)
        return({'val_loss':loss, 'val_acc': acc})

    def validation_epoch_end(self, outputs):
        batch_losses = [x['val_loss'] for x in outputs]
        epoch_loss = torch.stack(batch_losses).mean()
        batch_accs = [x['val_acc'] for x in outputs]
        epoch_acc = torch.stack(batch_accs).mean()
        return({'val_loss': epoch_loss.item(), 'val_acc' : epoch_acc.item()})
    def epoch_end(self, epoch,result):
        print("Epoch [{}], val_loss: {:.4f}, val_acc: {:.4f}".format(epoch, result['val_loss'], result['val_acc']))


model = MnistModel()
````

````python
def evaluate(model, val_loader):
    outputs = [model.validation_step(batch) for batch in val_loader]
    return(model.validation_epoch_end(outputs))

def fit(epochs, lr, model, train_loader, val_loader, opt_func = torch.optim.SGD):
    history = []
    optimizer = opt_func(model.parameters(), lr)
    for epoch in range(epochs):

        ## Training Phas
        for batch in train_loader:
            loss = model.training_step(batch)
            loss.backward()
            optimizer.step()
            optimizer.zero_grad()

        ## Validation phase
        result = evaluate(model, val_loader)
        model.epoch_end(epoch, result)
        history.append(result)
    return(history)
````

**optimizer.step() e loss.backward():** O algoritmo calcula o erro e ajusta os pesos do modelo para diminuir esse erro no futuro.
