## 34 - Qual o valor médio de frete cobrado por estado?

```python
# Baixa o dataset do Kaggle e carrega o DataFrame de pedidos
path = kagglehub.dataset_download("olistbr/brazilian-ecommerce")
orders_df = pd.read_csv(path + '/olist_orders_dataset.csv')
```

```python
# Carrega o DataFrame de itens de pedido
order_items_df = pd.read_csv(path + '/olist_order_items_dataset.csv')
```

```python
# Carrega o DataFrame de clientes
customers_df = pd.read_csv(path + '/olist_customers_dataset.csv')
```

```python
# Mescla os DataFrames de itens de pedido e pedidos
merged_orders_items_df = pd.merge(order_items_df, orders_df, on='order_id', how='inner')
```

```python
# Mescla o DataFrame combinado com o DataFrame de clientes
final_merged_df = pd.merge(merged_orders_items_df, customers_df, on='customer_id', how='inner')
```

```python
# Agrupa por estado do cliente e calcula o valor médio do frete
average_freight_per_state = final_merged_df.groupby('customer_state')['freight_value'].mean().reset_index()
```

```python
# Exibe o valor médio de frete por estado
print(average_freight_per_state)
```
