# === EXERCÍCIO 1: ANÁLISE BÁSICA DE SOBREVIVÊNCIA ===

## Objetivo
Criar uma função que analisa a taxa de sobrevivência dos passageiros do Titanic usando **apenas Python puro** (sem bibliotecas externas).

## Requisitos
A função deve:
- Receber uma **lista de dicionários** com dados dos passageiros
- Retornar um **dicionário** contendo:
  - `'total'`: número total de passageiros (tipo `int`)
  - `'sobreviventes'`: número de sobreviventes (tipo `int`)
  - `'taxa'`: taxa de sobrevivência em porcentagem (tipo `float`, arredondado para 2 casas decimais)

## Dicas
```python
# Como contar elementos em uma lista
total = len(lista)

# Como contar elementos que atendem uma condição
count = sum(1 for item in lista if item['campo'] == valor)

# Como calcular porcentagem
percentual = (parte / total) * 100

# Como arredondar para 2 casas decimais
arredondado = round(numero, 2)
```

## Exemplo
```python
dados = [
    {'Survived': 0},
    {'Survived': 1},
    {'Survived': 1}
]
resultado = analisar_sobrevivencia(dados)
print(resultado)
# Saída esperada: {'total': 3, 'sobreviventes': 2, 'taxa': 66.67}
```

````python
def analisar_sobrevivencia(dados):
    """
    Analisa taxa de sobrevivência do Titanic.

    Args:
        dados (list): Lista de dicionários com dados dos passageiros.
                     Cada dicionário deve ter a chave 'Survived' (0 ou 1).

    Returns:
        dict: Dicionário com as chaves:
              - 'total' (int): Total de passageiros
              - 'sobreviventes' (int): Número de sobreviventes
              - 'taxa' (float): Taxa de sobrevivência em %

    Exemplo:
        >>> analisar_sobrevivencia([{'Survived': 1}, {'Survived': 0}])
        {'total': 2, 'sobreviventes': 1, 'taxa': 50.0}
    """
    # === SEU CÓDIGO AQUI ===

    # armazena o tamanho do dicionário passado como parâmetro na função
    # inicializa a variável que será usada posteriormente
    total = len(dados)
    sobreviventes = 0

    # percorremos o dicionário 'dados' armazenando cada item em 'passageiros'
    # o if, fará a contagem de 'sobreviventes', caso 'Survived'=1, contabiliza 1 sobrevivente
    for passageiros in dados:
        if passageiros['Survived'] == 1:
            sobreviventes += 1

    # agora calculamos a taxa de sobreviventes e arredondamos para o maior inteiro mais prox em duas casas
    taxa = (sobreviventes / total) * 100
    taxa = round(taxa, 2)

    # adicionamos cada resultado em um dicionário e retornamos na função
    resultado = {
        'total' : total,
        'sobreviventes' : sobreviventes,
        'taxa' : taxa
    }
    return resultado

    # === FIM DO SEU CÓDIGO ===
````
