# Questão 1: Preparação e Limpeza de Dados (8 pontos)

## Objetivo
Preparar os dados para análise, tratando problemas de qualidade.

## Tarefa
Implemente a função `preparar_dados()` que:

1. **Combina os DataFrames** de vendas e clientes (merge)
2. **Trata valores missing:**
   - Remove linhas onde `valor` é NaN
   - Documenta quantas linhas foram removidas
3. **Cria novas colunas derivadas:**
   - `valor_total` = valor × quantidade
   - `mes` = mês extraído da data
   - `faixa_etaria` = categorização da idade:
     - 'Jovem': < 30
     - 'Adulto': 30-50
     - 'Senior': > 50
4. **Retorna dicionário com:**
   - `df_completo`: DataFrame limpo e enriquecido
   - `relatorio_limpeza`: dict com estatísticas da limpeza

## Exemplo do Relatório
```python
{
    'linhas_originais': 500,
    'linhas_removidas': 25,
    'linhas_finais': 475,
    'percentual_removido': 5.0
}
```

## Dicas
- Use `pd.merge()` para combinar os DataFrames
- Use `.dropna()` para remover missing
- Use `.apply()` ou condições para criar faixa_etaria
- Use `.dt.month` para extrair mês da data

**Tempo estimado:** 15 minutos

````python
def preparar_dados(df_vendas, df_clientes):
    """
    Prepara e limpa dados para análise.

    Args:
        df_vendas (pd.DataFrame): DataFrame de vendas
        df_clientes (pd.DataFrame): DataFrame de clientes

    Returns:
        dict: Dicionário com df_completo e relatorio_limpeza
    """
    # === SEU CÓDIGO AQUI ===

    # Combina os DataFrames
    df_completo = pd.merge(df_vendas, df_clientes, on='cliente_id', how='left')

    # Trata valores missing
    linhas_originais = df_completo.shape[0]
    df_completo.dropna(subset=['valor'], inplace=True)
    linhas_finais = df_completo.shape[0]
    linhas_removidas = linhas_originais - linhas_finais
    percentual_removido = (linhas_removidas / linhas_originais) * 100

    relatorio_limpeza = {
        'linhas_originais': linhas_originais,
        'linhas_removidas': linhas_removidas,
        'linhas_finais': linhas_finais,
        'percentual_removido': percentual_removido
    }

    # Cria novas colunas derivadas
    df_completo['valor_total'] = df_completo['valor'] * df_completo['quantidade']
    df_completo['mes'] = df_completo['data'].dt.month

    def categorizar_idade(idade):
        if idade < 30:
            return 'Jovem'
        elif 30 <= idade <= 50:
            return 'Adulto'
        else:
            return 'Senior'

    df_completo['faixa_etaria'] = df_completo['idade'].apply(categorizar_idade)

    return {
        'df_completo': df_completo,
        'relatorio_limpeza': relatorio_limpeza
    }

    # === FIM DO SEU CÓDIGO ===
````

````python
# === TESTE Q1 ===

print("=" * 70)
print("QUESTÃO 1: PREPARAÇÃO E LIMPEZA DE DADOS")
print("=" * 70)

# Preparar dados
resultado_prep = preparar_dados(df_vendas, df_clientes)
df_completo = resultado_prep['df_completo']
relatorio = resultado_prep['relatorio_limpeza']

print("\n=== RELATÓRIO DE LIMPEZA ===")
for chave, valor in relatorio.items():
    if 'percentual' in chave:
        print(f"{chave}: {valor:.2f}%")
    else:
        print(f"{chave}: {valor}")

print("\n=== AMOSTRA DOS DADOS LIMPOS ===")
print(df_completo.head())

print("\n=== INFORMAÇÕES DO DATASET ===")
print(f"Shape: {df_completo.shape}")
print(f"\nColunas: {list(df_completo.columns)}")

print("\n=== VERIFICAÇÕES ===")
print(f"✓ Missing em 'valor': {df_completo['valor'].isna().sum()}")
print(f"✓ Coluna 'valor_total' criada: {'valor_total' in df_completo.columns}")
print(f"✓ Coluna 'mes' criada: {'mes' in df_completo.columns}")
print(f"✓ Coluna 'faixa_etaria' criada: {'faixa_etaria' in df_completo.columns}")

print("\n=== DISTRIBUIÇÃO POR FAIXA ETÁRIA ===")
print(df_completo['faixa_etaria'].value_counts())

# Verificar testes
grader.check("q1")

print("\n" + "=" * 70)
print("FIM DA QUESTÃO 1")
print("=" * 70)
````

# Questão 2: Análise de Vendas Avançada (10 pontos)

## Objetivo
Realizar análise detalhada usando groupby, agregações e pivot tables.

## Tarefa
Implemente a função `analise_vendas_avancada()` que retorna dict com:

1. **por_regiao**: DataFrame com análise por região
   - total_vendas: soma de valor_total
   - num_transacoes: contagem de vendas
   - ticket_medio: média de valor_total
   - quantidade_media: média de quantidade

2. **por_categoria**: DataFrame com análise por categoria
   - Mesmas métricas que por_regiao

3. **pivot_regiao_categoria**: Pivot table
   - Linhas: região
   - Colunas: categoria
   - Valores: soma de valor_total

4. **top_clientes**: DataFrame com top 10 clientes por valor_total
   - Colunas: cliente_id, nome, total_gasto, num_compras

5. **metricas_gerais**: dict com:
   - receita_total
   - num_vendas_total
   - ticket_medio_geral
   - regiao_mais_lucrativa
   - categoria_mais_vendida

## Dicas
- Use `groupby().agg()` para múltiplas agregações
- Use `pivot_table()` para tabela cruzada
- Use `.nlargest()` para top clientes

**Tempo estimado:** 20 minutos

````python
def analise_vendas_avancada(df):
    """
    Análise avançada de vendas com groupby e pivot.

    Args:
        df (pd.DataFrame): DataFrame completo

    Returns:
        dict: Dicionário com todas as análises
    """
    # === SEU CÓDIGO AQUI ===

    # Análise por região
    por_regiao = df.groupby('regiao').agg(
        total_vendas=('valor_total', 'sum'),
        num_transacoes=('venda_id', 'count'),
        ticket_medio=('valor_total', 'mean'),
        quantidade_media=('quantidade', 'mean')
    ).reset_index()

    # Análise por categoria
    por_categoria = df.groupby('categoria').agg(
        total_vendas=('valor_total', 'sum'),
        num_transacoes=('venda_id', 'count'),
        ticket_medio=('valor_total', 'mean'),
        quantidade_media=('quantidade', 'mean')
    ).reset_index()

    # Pivot table: região x categoria
    pivot_regiao_categoria = df.pivot_table(
        index='regiao',
        columns='categoria',
        values='valor_total',
        aggfunc='sum',
        fill_value=0
    )

    # Top 10 clientes
    top_clientes = df.groupby(['cliente_id', 'nome']).agg(
        total_gasto=('valor_total', 'sum'),
        num_compras=('venda_id', 'count')
    ).reset_index().nlargest(10, 'total_gasto')

    # Métricas gerais
    receita_total = df['valor_total'].sum()
    num_vendas_total = df.shape[0]
    ticket_medio_geral = df['valor_total'].mean()
    regiao_mais_lucrativa = por_regiao.loc[por_regiao['total_vendas'].idxmax(), 'regiao']
    categoria_mais_vendida = por_categoria.loc[por_categoria['num_transacoes'].idxmax(), 'categoria']

    metricas_gerais = {
        'receita_total': receita_total,
        'num_vendas_total': num_vendas_total,
        'ticket_medio_geral': ticket_medio_geral,
        'regiao_mais_lucrativa': regiao_mais_lucrativa,
        'categoria_mais_vendida': categoria_mais_vendida
    }

    return {
        'por_regiao': por_regiao,
        'por_categoria': por_categoria,
        'pivot_regiao_categoria': pivot_regiao_categoria,
        'top_clientes': top_clientes,
        'metricas_gerais': metricas_gerais
    }

    # === FIM DO SEU CÓDIGO ===
````

````python
# === TESTE Q2 ===

print("=" * 70)
print("QUESTÃO 2: ANÁLISE DE VENDAS AVANÇADA")
print("=" * 70)

# Realizar análise
analise = analise_vendas_avancada(df_completo)

print("\n=== 1. ANÁLISE POR REGIÃO ===")
print(analise['por_regiao'].round(2))

print("\n=== 2. ANÁLISE POR CATEGORIA ===")
print(analise['por_categoria'].round(2))

print("\n=== 3. PIVOT TABLE: REGIÃO × CATEGORIA ===")
print(analise['pivot_regiao_categoria'].round(2))

print("\n=== 4. TOP 10 CLIENTES ===")
print(analise['top_clientes'].head(10).round(2))

print("\n=== 5. MÉTRICAS GERAIS ===")
metricas = analise['metricas_gerais']
print(f"Receita Total: R$ {metricas['receita_total']:,.2f}")
print(f"Número de Vendas: {metricas['num_vendas_total']}")
print(f"Ticket Médio: R$ {metricas['ticket_medio_geral']:.2f}")
print(f"Região Mais Lucrativa: {metricas['regiao_mais_lucrativa']}")
print(f"Categoria Mais Vendida: {metricas['categoria_mais_vendida']}")

# Visualizações
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Gráfico 1: Vendas por Região
analise['por_regiao'].plot(x='regiao', y='total_vendas', kind='bar', ax=axes[0], color='steelblue')
axes[0].set_title('Total de Vendas por Região', fontweight='bold')
axes[0].set_xlabel('Região')
axes[0].set_ylabel('Total de Vendas (R$)')
axes[0].tick_params(axis='x', rotation=45)

# Gráfico 2: Vendas por Categoria
analise['por_categoria'].plot(x='categoria', y='total_vendas', kind='bar', ax=axes[1], color='coral')
axes[1].set_title('Total de Vendas por Categoria', fontweight='bold')
axes[1].set_xlabel('Categoria')
axes[1].set_ylabel('Total de Vendas (R$)')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# Verificar testes
grader.check("q2")

print("\n" + "=" * 70)
print("FIM DA QUESTÃO 2")
print("=" * 70)
````

# Questão 3: Análise Estatística Completa (10 pontos)

## Objetivo
Realizar análise estatística descritiva e identificar outliers.

## Tarefa
Implemente a função `analise_estatistica_completa()` que retorna dict com:

1. **estatisticas**: DataFrame com estatísticas para colunas numéricas
   - Colunas: media, mediana, desvio_padrao, cv, min, max, iqr
   - Linhas: cada variável numérica

2. **outliers**: dict onde chave é nome da coluna, valor é:
   - Series com os valores outliers
   - Use regra IQR

3. **distribuicoes**: dict com classificação da distribuição:
   - 'simetrica' ou 'assimetrica' para cada variável numérica
   - Base: |media - mediana| < 0.1 × desvio_padrao

4. **visualizacoes**: matplotlib Figure com 2×2 subplots:
   - [0,0]: Histogramas de valor_total
   - [0,1]: Boxplot de valor_total
   - [1,0]: Histograma de quantidade
   - [1,1]: Boxplot de quantidade

**Tempo estimado:** 15 minutos

````python
def analise_estatistica_completa(df):
    """
    Análise estatística completa com outliers e visualizações.

    Args:
        df (pd.DataFrame): DataFrame

    Returns:
        dict: Análise estatística completa
    """
    # === SEU CÓDIGO AQUI ===

    numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

    # Estatísticas descritivas
    stats_dict = {}
    for col in numeric_cols:
        mean_val = df[col].mean()
        median_val = df[col].median()
        std_val = df[col].std()
        min_val = df[col].min()
        max_val = df[col].max()
        q1 = df[col].quantile(0.25)
        q3 = df[col].quantile(0.75)
        iqr_val = q3 - q1
        cv_val = (std_val / mean_val) if mean_val != 0 else np.nan

        stats_dict[col] = {
            'media': mean_val,
            'mediana': median_val,
            'desvio_padrao': std_val,
            'cv': cv_val,
            'min': min_val,
            'max': max_val,
            'iqr': iqr_val
        }
    estatisticas = pd.DataFrame.from_dict(stats_dict, orient='index')

    # Outliers
    outliers = {}
    for col in numeric_cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        col_outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)][col]
        outliers[col] = col_outliers


    # Classificação das distribuições
    distribuicoes = {}
    for col in numeric_cols:
        mean_val = df[col].mean()
        median_val = df[col].median()
        std_val = df[col].std()

        if pd.isna(std_val) or std_val == 0:
             distribuicoes[col] = 'não aplicavel'
        elif abs(mean_val - median_val) < 0.1 * std_val:
            distribuicoes[col] = 'simetrica'
        else:
            distribuicoes[col] = 'assimetrica'

    # Visualizações
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))

    sns.histplot(df['valor_total'], kde=True, ax=axes[0, 0])
    axes[0, 0].set_title('Histograma de Valor Total')

    sns.boxplot(x=df['valor_total'], ax=axes[0, 1])
    axes[0, 1].set_title('Boxplot de Valor Total')

    sns.histplot(df['quantidade'], kde=True, ax=axes[1, 0])
    axes[1, 0].set_title('Histograma de Quantidade')

    sns.boxplot(x=df['quantidade'], ax=axes[1, 1])
    axes[1, 1].set_title('Boxplot de Quantidade')

    plt.tight_layout()

    return {
        'estatisticas': estatisticas,
        'outliers': outliers,
        'distribuicoes': distribuicoes,
        'visualizacoes': fig
    }

    # === FIM DO SEU CÓDIGO ===
````

````python
# === TESTE Q3 ===

print("=" * 70)
print("QUESTÃO 3: ANÁLISE ESTATÍSTICA COMPLETA")
print("=" * 70)

# Realizar análise
analise_stats = analise_estatistica_completa(df_completo)

print("\n=== 1. ESTATÍSTICAS DESCRITIVAS ===")
print(analise_stats['estatisticas'].round(2))

print("\n=== 2. OUTLIERS DETECTADOS ===")
for col, outlier_vals in analise_stats['outliers'].items():
    if len(outlier_vals) > 0:
        print(f"\n{col}: {len(outlier_vals)} outliers")
        print(f"  Valores: {sorted(outlier_vals.values)[:5]}..." if len(outlier_vals) > 5 else f"  Valores: {sorted(outlier_vals.values)}")

print("\n=== 3. CLASSIFICAÇÃO DAS DISTRIBUIÇÕES ===")
for col, tipo in analise_stats['distribuicoes'].items():
    print(f"{col}: {tipo}")

print("\n=== 4. VISUALIZAÇÕES ===")
analise_stats['visualizacoes']
plt.show()

# Verificar testes
grader.check("q3")

print("\n" + "=" * 70)
print("FIM DA QUESTÃO 3")
print("=" * 70)
````

# Questão 4: Dashboard Final de EDA (12 pontos)

## Objetivo
Criar dashboard visual completo com análise de correlações e insights.

## Tarefa
Implemente a função `dashboard_final_eda()` que retorna dict com:

1. **correlacoes**: dict com:
   - `matriz`: DataFrame com matriz de correlação
   - `pares_fortes`: lista de tuplas (var1, var2, corr) onde |corr| >= 0.5
   - `heatmap`: matplotlib Figure com heatmap

2. **insights**: dict com:
   - `melhor_regiao`: tupla (nome, receita)
   - `melhor_categoria`: tupla (nome, num_vendas)
   - `perfil_cliente_ideal`: dict com faixa_etaria e tempo_cliente mais lucrativos
   - `periodo_pico`: mês com mais vendas

3. **dashboard**: matplotlib Figure 2×3 com:
   - [0,0]: Scatter plot valor vs quantidade
   - [0,1]: Boxplot valor_total por faixa_etaria
   - [0,2]: Bar chart vendas por mês
   - [1,0]: Scatter valor_total vs idade (com cor por região)
   - [1,1]: Violin plot valor_total por região
   - [1,2]: Texto com resumo executivo

4. **recomendacoes**: lista de 5 recomendações estratégicas
   baseadas na análise

**Tempo estimado:** 10 minutos

````python
def dashboard_final_eda(df):
    """
    Dashboard final com correlações, insights e recomendações.

    Args:
        df (pd.DataFrame): DataFrame completo

    Returns:
        dict: Dashboard completo
    """
    # === SEU CÓDIGO AQUI ===

    # 1. Correlações
    numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
    matriz_correlacao = df[numeric_cols].corr()

    pares_fortes = []
    # Percorre o triângulo superior da matriz de correlação.
    for i in range(len(matriz_correlacao.columns)):
        for j in range(i + 1, len(matriz_correlacao.columns)):
            col1 = matriz_correlacao.columns[i]
            col2 = matriz_correlacao.columns[j]
            corr_val = matriz_correlacao.iloc[i, j]
            if abs(corr_val) >= 0.5:
                pares_fortes.append((col1, col2, corr_val))

    fig_heatmap, ax_heatmap = plt.subplots(figsize=(10, 8))
    sns.heatmap(matriz_correlacao, annot=True, cmap='coolwarm', fmt=".2f", ax=ax_heatmap)
    ax_heatmap.set_title('Heatmap de Correlação', fontweight='bold')

    correlacoes = {
        'matriz': matriz_correlacao,
        'pares_fortes': pares_fortes,
        'heatmap': fig_heatmap
    }

    # Insights
    # Calcular métricas gerais primeiro
    receita_total = df['valor_total'].sum()
    num_vendas_total = df.shape[0]
    ticket_medio_geral = df['valor_total'].mean()

    # Melhor Região (baseado em receita total)
    vendas_por_regiao = df.groupby('regiao')['valor_total'].sum().reset_index()
    melhor_regiao_data = vendas_por_regiao.loc[vendas_por_regiao['valor_total'].idxmax()]
    melhor_regiao = (melhor_regiao_data['regiao'], melhor_regiao_data['valor_total'])

    # Melhor Categoria (baseado em número de vendas)
    vendas_por_categoria = df.groupby('categoria')['venda_id'].count().reset_index()
    melhor_categoria_data = vendas_por_categoria.loc[vendas_por_categoria['venda_id'].idxmax()]
    melhor_categoria = (melhor_categoria_data['categoria'], melhor_categoria_data['venda_id'])

    # Perfil Cliente Ideal (mais lucrativo por faixa etaria e tempo de cliente)
    perfil_lucrativo_faixa_etaria = df.groupby('faixa_etaria')['valor_total'].sum().idxmax()
    perfil_lucrativo_tempo_cliente = df.groupby('tempo_cliente_meses')['valor_total'].sum().idxmax()

    perfil_cliente_ideal = {
        'faixa_etaria': perfil_lucrativo_faixa_etaria,
        'tempo_cliente_meses': perfil_lucrativo_tempo_cliente
    }

    # Período Pico (mês com mais vendas)
    vendas_por_mes = df.groupby('mes')['venda_id'].count().reset_index()
    periodo_pico = vendas_por_mes.loc[vendas_por_mes['venda_id'].idxmax(), 'mes']

    # Agora popular o dicionário insights
    insights = {
        'receita_total': receita_total,
        'num_vendas_total': num_vendas_total,
        'ticket_medio_geral': ticket_medio_geral,
        'melhor_regiao': melhor_regiao,
        'melhor_categoria': melhor_categoria,
        'perfil_cliente_ideal': perfil_cliente_ideal,
        'periodo_pico': periodo_pico
    }

    # 3. Dashboard Visual
    fig_dashboard, axes_dashboard = plt.subplots(2, 3, figsize=(18, 12))

    # [0,0]: Scatter plot valor vs quantidade
    sns.scatterplot(data=df, x='valor', y='quantidade', ax=axes_dashboard[0, 0])
    axes_dashboard[0, 0].set_title('Valor vs Quantidade', fontweight='bold')

    # [0,1]: Boxplot valor_total por faixa_etaria
    sns.boxplot(data=df, x='faixa_etaria', y='valor_total', ax=axes_dashboard[0, 1])
    axes_dashboard[0, 1].set_title('Valor Total por Faixa Etária', fontweight='bold')

    # [0,2]: Bar chart vendas por mês
    sns.countplot(data=df, x='mes', ax=axes_dashboard[0, 2], palette='viridis')
    axes_dashboard[0, 2].set_title('Número de Vendas por Mês', fontweight='bold')
    axes_dashboard[0, 2].set_xlabel('Mês')
    axes_dashboard[0, 2].set_ylabel('Número de Vendas')


    # [1,0]: Scatter valor_total vs idade (com cor por região)
    sns.scatterplot(data=df, x='idade', y='valor_total', hue='regiao', ax=axes_dashboard[1, 0])
    axes_dashboard[1, 0].set_title('Valor Total vs Idade (por Região)', fontweight='bold')

    # [1,1]: Violin plot valor_total por região
    sns.violinplot(data=df, x='regiao', y='valor_total', ax=axes_dashboard[1, 1], palette='pastel')
    axes_dashboard[1, 1].set_title('Distribuição de Valor Total por Região', fontweight='bold')
    axes_dashboard[1, 1].tick_params(axis='x', rotation=45)

    # [1,2]: Texto com resumo executivo
    resumo_executivo = f"""
    Resumo Executivo de Vendas:

    Receita Total: R$ {insights['receita_total']:,.2f}
    Total de Vendas: {insights['num_vendas_total']}
    Ticket Médio Geral: R$ {insights['ticket_medio_geral']:.2f}

    Região Mais Lucrativa: {insights['melhor_regiao'][0]}
    Categoria Mais Vendida: {insights['melhor_categoria'][0]}
    Período de Pico de Vendas: Mês {insights['periodo_pico']}

    Perfil de Cliente Ideal (Mais Lucrativo):
    - Faixa Etária: {insights['perfil_cliente_ideal']['faixa_etaria']}
    - Tempo de Cliente: {insights['perfil_cliente_ideal']['tempo_cliente_meses']} meses

    Correlações Fortes (|r| >= 0.5):
    """
    if pares_fortes:
        for var1, var2, corr in pares_fortes:
             resumo_executivo += f"\n  - {var1} ↔ {var2}: {corr:+.3f}"
    else:
        resumo_executivo += "\n  - Nenhum par com correlação >= 0.5"


    axes_dashboard[1, 2].text(0.05, 0.95, resumo_executivo,
                              verticalalignment='top', horizontalalignment='left',
                              wrap=True, fontsize=10, bbox={'facecolor': 'white', 'alpha': 0.5, 'pad': 10})
    axes_dashboard[1, 2].set_title('Resumo Executivo e Insights', fontweight='bold')
    axes_dashboard[1, 2].axis('off') # Hide axes for the text plot


    plt.tight_layout()
    # plt.show() # Do not show plot here, return the figure object


    # 4. Recomendações Estratégicas
    recomendacoes = [
        f"Investir em campanhas de marketing direcionadas para a região '{insights['melhor_regiao'][0]}', que é a mais lucrativa.",
        f"Analisar a fundo a categoria '{insights['melhor_categoria'][0]}' para entender os fatores de sucesso e replicá-los em outras categorias.",
        f"Criar programas de fidelidade ou ofertas especiais para clientes com perfil '{insights['perfil_cliente_ideal']['faixa_etaria']}' e com '{insights['perfil_cliente_ideal']['tempo_cliente_meses']}' meses de tempo de cliente, pois representam o perfil mais lucrativo.",
        f"Planejar reforços de estoque e campanhas promocionais para o Mês {insights['periodo_pico']}, que é o período de pico de vendas.",
        "Investigar os outliers de 'valor_total' para entender se representam fraudes, erros de registro ou vendas de alto valor que podem indicar oportunidades para produtos premium."
    ]


    return {
        'correlacoes': correlacoes,
        'insights': insights,
        'dashboard': fig_dashboard, # Return the figure object
        'recomendacoes': recomendacoes
    }

    # === FIM DO SEU CÓDIGO ===
````

````python
# === TESTE Q4 ===

print("=" * 70)
print("QUESTÃO 4: DASHBOARD FINAL DE EDA")
print("=" * 70)

# Criar dashboard
dashboard = dashboard_final_eda(df_completo)

print("\n=== 1. CORRELAÇÕES ===")
print(f"\nPares com correlação forte (|r| >= 0.5):")
if dashboard['correlacoes']['pares_fortes']:
    for v1, v2, corr in dashboard['correlacoes']['pares_fortes']:
        print(f"  {v1} ↔ {v2}: {corr:+.3f}")
else:
    print("  Nenhum par com correlação >= 0.5")

print("\n=== HEATMAP DE CORRELAÇÃO ===")
dashboard['correlacoes']['heatmap']
plt.show()

print("\n=== 2. INSIGHTS PRINCIPAIS ===")
insights = dashboard['insights']
print(f"\nMelhor Região: {insights['melhor_regiao'][0]}")
print(f"  Receita: R$ {insights['melhor_regiao'][1]:,.2f}")

print(f"\nMelhor Categoria: {insights['melhor_categoria'][0]}")
print(f"  Vendas: {insights['melhor_categoria'][1]}")

print(f"\nPerfil Cliente Ideal:")
print(f"  Faixa Etária: {insights['perfil_cliente_ideal']['faixa_etaria']}")
print(f"  Tempo de Cliente: {insights['perfil_cliente_ideal']['tempo_cliente_meses']} meses")

print(f"\nPeríodo Pico: Mês {insights['periodo_pico']}")

print("\n=== 3. DASHBOARD EXECUTIVO ===")
dashboard['dashboard']
plt.show()

print("\n=== 4. RECOMENDAÇÕES ESTRATÉGICAS ===")
for i, rec in enumerate(dashboard['recomendacoes'], 1):
    print(f"{i}. {rec}")

# Verificar testes
grader.check("q4")

print("\n" + "=" * 70)
print("FIM DA QUESTÃO 4")
print("=" * 70)
````
