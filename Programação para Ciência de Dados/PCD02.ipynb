# === EXERCÍCIO 1: ESTRUTURAS ANINHADAS ===

## Objetivo
Criar uma **matriz de sobrevivência** usando listas aninhadas que organize os dados por classe e sexo.

## Requisitos
A função deve:
- Receber uma **lista de dicionários** com dados dos passageiros
- Retornar uma **matriz 3x2** (lista de listas):
  - Linhas: Classes 1, 2 e 3
  - Colunas: Male (coluna 0) e Female (coluna 1)
  - Valores: **Quantidade de sobreviventes** em cada categoria

## Estrutura da Matriz
```python
[
  [male_class1, female_class1],  # Linha 0: Classe 1
  [male_class2, female_class2],  # Linha 1: Classe 2
  [male_class3, female_class3]   # Linha 2: Classe 3
]
```

## Dicas
```python
# Criar matriz 3x2 preenchida com zeros
matriz = [[0 for j in range(2)] for i in range(3)]

# Acessar posição: linha (classe-1), coluna (0=male, 1=female)
matriz[classe - 1][0 if sexo == 'male' else 1] += 1

# Filtrar sobreviventes
if passageiro['Survived'] == 1:
    # incrementar contador
```

## Exemplo
```python
dados = [
    {'Pclass': 1, 'Sex': 'male', 'Survived': 1},
    {'Pclass': 1, 'Sex': 'female', 'Survived': 1},
    {'Pclass': 2, 'Sex': 'male', 'Survived': 0}
]
resultado = criar_matriz_sobrevivencia(dados)
# Resultado esperado: [[1, 1], [0, 0], [0, 0]]
```

````python
def criar_matriz_sobrevivencia(dados):
    """
    Cria matriz de sobrevivência organizada por classe e sexo.

    Args:
        dados (list): Lista de dicionários com dados dos passageiros.
                     Cada dicionário deve ter: 'Pclass', 'Sex', 'Survived'

    Returns:
        list: Matriz 3x2 com contagem de sobreviventes.
              Linhas = classes (1, 2, 3)
              Colunas = sexos (male=0, female=1)

    Exemplo:
        >>> criar_matriz_sobrevivencia([{'Pclass': 1, 'Sex': 'male', 'Survived': 1}])
        [[1, 0], [0, 0], [0, 0]]
    """
    # === SEU CÓDIGO AQUI ===
    matriz = [[0 for j in range(2)] for i in range(3)]

    for passageiro in dados:

        if passageiro['Survived'] == 1:
            classe = passageiro['Pclass']
            sexo = passageiro['Sex']

            matriz[classe - 1][0 if sexo == 'male' else 1] += 1

    return matriz
    # === FIM DO SEU CÓDIGO ===
````

````python
# Testar sua função com o dataset
matriz = criar_matriz_sobrevivencia(titanic_data)

print("\n=== MATRIZ DE SOBREVIVÊNCIA ===")
print("\nClasses (linhas) x Sexo (colunas)")
print("Colunas: [Male, Female]\n")

classes = ['Classe 1', 'Classe 2', 'Classe 3']
for i, classe in enumerate(classes):
    print(f"{classe}: {matriz[i]}")
    print(f"  - Homens sobreviventes: {matriz[i][0]}")
    print(f"  - Mulheres sobreviventes: {matriz[i][1]}")
    print()
````

# === EXERCÍCIO 2: COMPREHENSIONS E COLLECTIONS ===

## Objetivo
Usar **dictionary comprehension** e **collections** para análise agregada por classe.

## Requisitos
A função deve:
- Receber uma **lista de dicionários** com dados dos passageiros
- Retornar um **dicionário** onde:
  - Chaves: Número da classe (1, 2, 3)
  - Valores: Dicionário com estatísticas da classe:
    - `'passageiros'`: Quantidade total de passageiros (int)
    - `'tarifa_media'`: Tarifa média da classe (float)
    - `'taxa_sobrevivencia'`: Taxa de sobrevivência em % (float)

## Dicas
```python
# Usar collections.defaultdict para agrupar
from collections import defaultdict
grupos = defaultdict(list)

# Dictionary comprehension para criar resultado
resultado = {
    classe: {
        'passageiros': len(passageiros),
        'tarifa_media': sum(p['Fare'] for p in passageiros) / len(passageiros),
        'taxa_sobrevivencia': ...
    }
    for classe, passageiros in grupos.items()
}

# Calcular taxa: (sobreviventes / total) * 100
```

## Importante
- Todas as classes (1, 2, 3) devem estar no dicionário retornado
- Se uma classe não tem passageiros, use valores 0

## Exemplo
```python
dados = [
    {'Pclass': 1, 'Fare': 50.0, 'Survived': 1},
    {'Pclass': 1, 'Fare': 30.0, 'Survived': 0}
]
resultado = analisar_por_classe(dados)
# resultado[1] => {'passageiros': 2, 'tarifa_media': 40.0, 'taxa_sobrevivencia': 50.0}
```

````python
from collections import defaultdict

def analisar_por_classe(dados):
    """
    Analisa estatísticas de passageiros por classe usando comprehensions.

    Args:
        dados (list): Lista de dicionários com dados dos passageiros.
                     Cada dicionário deve ter: 'Pclass', 'Fare', 'Survived'

    Returns:
        dict: Dicionário com chaves 1, 2, 3 (classes).
              Cada valor é um dict com 'passageiros', 'tarifa_media', 'taxa_sobrevivencia'

    Exemplo:
        >>> analisar_por_classe([{'Pclass': 1, 'Fare': 50, 'Survived': 1}])
        {1: {'passageiros': 1, 'tarifa_media': 50.0, 'taxa_sobrevivencia': 100.0}, ...}
    """
    # === SEU CÓDIGO AQUI ===
    grupos = defaultdict(list)
    for passageiro in dados:
        grupos[passageiro['Pclass']].append(passageiro)

    return {
        classe: {
            'passageiros': len(passageiros),
            'tarifa_media': sum(p['Fare'] for p in passageiros) / len(passageiros) if passageiros else 0,
            'taxa_sobrevivencia': (sum(p['Survived'] for p in passageiros) / len(passageiros) * 100) if passageiros else 0
        }
        for classe in [1, 2, 3]
        for passageiros in [grupos.get(classe, [])]
    }
    # === FIM DO SEU CÓDIGO ===
````

````python
# Testar sua função com o dataset
analise = analisar_por_classe(titanic_data)

print("\n=== ANÁLISE POR CLASSE ===")
for classe in [1, 2, 3]:
    stats = analise[classe]
    print(f"\nClasse {classe}:")
    print(f"  - Total de passageiros: {stats['passageiros']}")
    print(f"  - Tarifa média: ${stats['tarifa_media']:.2f}")
    print(f"  - Taxa de sobrevivência: {stats['taxa_sobrevivencia']:.2f}%")
````

# === EXERCÍCIO 3: ORDENAÇÃO CUSTOMIZADA ===

## Objetivo
Implementar função de **ordenação customizada** usando `sorted()` com parâmetro `key`.

## Requisitos
A função deve:
- Receber:
  - Uma **lista de dicionários** com dados dos passageiros
  - Um **campo** para ordenação (string)
  - Um parâmetro **reverso** (bool, padrão False)
- Retornar uma **nova lista ordenada** (não modificar a original)
- Ordenar pelo campo especificado
- Se reverso=True, ordenar em ordem decrescente

## Dicas
```python
# Usar sorted() com key lambda
sorted(dados, key=lambda x: x[campo])

# Para ordem decrescente
sorted(dados, key=lambda x: x[campo], reverse=True)

# Lambda acessa valor do dicionário dinamicamente
campo = 'Age'
lambda x: x[campo]  # equivalente a lambda x: x['Age']
```

## Importante
- NÃO modificar a lista original
- Usar `sorted()`, não `.sort()`
- Funcionar com qualquer campo do dicionário

## Exemplo
```python
dados = [
    {'Name': 'Ana', 'Age': 30},
    {'Name': 'Bruno', 'Age': 20}
]
resultado = ordenar_passageiros(dados, 'Name')
# [{'Name': 'Ana', 'Age': 30}, {'Name': 'Bruno', 'Age': 20}]

resultado = ordenar_passageiros(dados, 'Name', reverso=True)
# [{'Name': 'Bruno', 'Age': 20}, {'Name': 'Ana', 'Age': 30}]
```

````python
def ordenar_passageiros(dados, campo, reverso=False):
    """
    Ordena passageiros por campo especificado.

    Args:
        dados (list): Lista de dicionários com dados dos passageiros
        campo (str): Nome do campo para ordenação (ex: 'Age', 'Fare', 'Name')
        reverso (bool): Se True, ordena em ordem decrescente

    Returns:
        list: Nova lista ordenada (não modifica a original)

    Exemplo:
        >>> dados = [{'Name': 'B', 'Age': 30}, {'Name': 'A', 'Age': 20}]
        >>> ordenar_passageiros(dados, 'Age')
        [{'Name': 'A', 'Age': 20}, {'Name': 'B', 'Age': 30}]
    """
    # === SEU CÓDIGO AQUI ===

    return sorted(dados, key=lambda x: x.get(campo, 0), reverse=reverso)

    # === FIM DO SEU CÓDIGO ===
````

# === EXERCÍCIO 4: OTIMIZAÇÃO COM DICIONÁRIO ===

## Objetivo
Implementar **busca eficiente** usando dicionário como índice (O(1) vs O(n)).

## Requisitos
A função deve:
- Receber:
  - Uma **lista de dicionários** com dados dos passageiros
  - Um **PassengerId** para buscar
- Retornar:
  - O **dicionário completo** do passageiro encontrado
  - `None` se não encontrar
- Usar **dicionário como índice** para busca O(1)

## Por que isso é eficiente?
```python
# Busca linear (ineficiente) - O(n)
for p in dados:
    if p['PassengerId'] == id:
        return p

# Busca com dicionário (eficiente) - O(1)
indice = {p['PassengerId']: p for p in dados}
return indice.get(id)
```
