1 - A partir do conjunto de dados:
- fazer um histograma dos valores de diâmetros com 10, 20, 30 e 300 divisões,
observe e explique as razões dos gráficos serem tão diferentes;
- fazer um histograma de densidade;

```python
import matplotlib.pyplot as plt
import pandas as pd

# Histograms for Diameter with different bin sizes
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
X['Diameter'].hist(bins=10)
plt.title('Diameter Histogram with 10 Bins')

plt.subplot(2, 2, 2)
X['Diameter'].hist(bins=20)
plt.title('Diameter Histogram with 20 Bins')

plt.subplot(2, 2, 3)
X['Diameter'].hist(bins=30)
plt.title('Diameter Histogram with 30 Bins')

plt.subplot(2, 2, 4)
X['Diameter'].hist(bins=300)
plt.title('Diameter Histogram with 300 Bins')

plt.tight_layout()
plt.show()

# Explanation for different histograms
print("Explanation:")
print("Histograms with different bin sizes show varying levels of detail in the data distribution.")
print("- With fewer bins (e.g., 10), the histogram provides a general overview of the data shape, grouping a wider range of values together.")
print("- As the number of bins increases (e.g., 300), the histogram reveals finer details and potential modes or gaps in the data that might not be visible with fewer bins.")
print("- However, too many bins can make the histogram look noisy and less smooth, potentially highlighting random fluctuations rather than underlying patterns.")

# Density histogram for Diameter
plt.figure(figsize=(6, 4))
X['Diameter'].plot(kind='hist', density=True, bins=30)
plt.title('Diameter Density Histogram')
plt.xlabel('Diameter')
plt.ylabel('Density')
plt.show()

# Histograms for Shell_weight with different bin sizes
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
X['Shell_weight'].hist(bins=10)
plt.title('Shell_weight Histogram with 10 Bins')

plt.subplot(2, 2, 2)
X['Shell_weight'].hist(bins=20)
plt.title('Shell_weight Histogram with 20 Bins')

plt.subplot(2, 2, 3)
X['Shell_weight'].hist(bins=30)
plt.title('Shell_weight Histogram with 30 Bins')

plt.subplot(2, 2, 4)
X['Shell_weight'].hist(bins=300)
plt.title('Shell_weight Histogram with 300 Bins')

plt.tight_layout()
plt.show()

# Density histogram for Shell_weight
plt.figure(figsize=(6, 4))
X['Shell_weight'].plot(kind='hist', density=True, bins=30)
plt.title('Shell_weight Density Histogram')
plt.xlabel('Shell_weight')
plt.ylabel('Density')
plt.show()
```
