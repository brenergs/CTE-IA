## 1 - Escolher uma variável quantitativa e uma qualitativa e caracterizar com moda, mediana e os quartis, além de média e variância.


```python
# Quantitative variable: Height
quantitativa = X['Height']
print('Media: ', quantitativa.mean())
print('Mediana: ', quantitativa.median())
print('Moda: ', quantitativa.mode())
print('Variância: ', quantitativa.var())
print('Quartil: ')
print(quantitativa.quantile([0.25, 0.5, 0.75]))

print('')

qualitativa = X['Sex']
print('Moda: ', qualitativa.mode())
```
