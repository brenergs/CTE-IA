# Ex2

- Extraia uma outra tabela com os espécimes que tem comprimento acima da média

- Calcula para estes a média do peso total

```python
decs.describe()
```

```python
print(decs['Length'].mean().round(2)) # média do tamanho


decs.loc[decs['Length']>decs['Length'].mean(), 'Whole_weight']
```

```python
# retorna a média do peso total daquelas linhas cujo comprimento está acima da média
decs.loc[decs['Length']>decs['Length'].mean(), ['Whole_weight', 'Shucked_weight']].mean()
```
