1 - Criar um classificador de floresta aleatória para o conjunto de dados de vinhos.

Mostrar a variação da acurácia quando se varia o número de árvores entre 5 e 50

```python
from sklearn.datasets import load_wine

# Carrega o dataset com os dados dos vinhos
wine = load_wine()
X, y = wine.data, wine.target

# Separa os dados entre conjunto de teste e de treino
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Cria lista para armazenar o número de arvores e a acurácia
lista_arvores = range(5, 51)
acuracia = []

# Percorre diferentes números de árvores e cria um classificador de árvores aleatórias
# Realiza o treino dos classificadores os avalia
for arvores in lista_arvores:
    classificador_aleatorio = RandomForestClassifier(n_estimators=arvores, random_state=42)
    classificador_aleatorio.fit(X_train, y_train)

    # Previsões do conjunto de teste
    y_pred = classificador_aleatorio.predict(X_test)

    # Calcula a acurácia dividindo o número de previsões corretas pelo seu total
    calculo_acuracia = (y_pred == y_test).sum() / len(y_test)

    acuracia.append(calculo_acuracia)


for n, acc in zip(lista_arvores, acuracia):
    print(f"Número de árvores: {n}, Acurácia: {acc:.4f}")
```
