## 1 - Carregar os dados em uma tabela.

```python
from ucimlrepo import fetch_ucirepo

# fetch dataset
wine_quality = fetch_ucirepo(id=186)

# data (as pandas dataframes)
X = wine_quality.data.features
y = wine_quality.data.targets

# metadata
print(wine_quality.metadata)

# variable information
print(wine_quality.variables)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.concat([X, y], axis=1)
```


## 2 - Fazer um gráfico de pontos de duas variáveis.

```python
plt.boxplot(df[['alcohol', 'quality']].values, tick_labels=['Álcool', 'Qualidade'])
plt.title('Qualidade do Vinho')
plt.grid(True)
plt.show()
```

## 3 - Histograma de 4 variáveis com 10 e com 40 divisões.

```python
variables_to_plot = ['alcohol', 'citric_acid', 'residual_sugar', 'density']

plt.figure(figsize=(15, 10))
for i, var in enumerate(variables_to_plot):
    plt.subplot(2, 2, i + 1)
    plt.hist(df[var], bins=10, edgecolor='black')
    plt.title(f'Histograma de {var.replace("_", " ").title()} (10 Divisões)')
    plt.grid(True)
plt.show()

plt.figure(figsize=(15, 10))
for i, var in enumerate(variables_to_plot):
    plt.subplot(2, 2, i + 1)
    plt.hist(df[var], bins=40, edgecolor='black')
    plt.title(f'Histograma de {var.replace("_", " ").title()} (40 Divisões)')
    plt.grid(True)
plt.show()
```

## 4 - Calcular média, mediana, moda e quartis para todas as variáveis.

```python
print('=========================Média===========================')
print(df.mean())
print('========================Mediana==========================')
print(df.median())
print('=========================Moda============================')
print(df.mode())
print('=========================Quartis=========================')
df.quantile([0.25, 0.5, 0.75])
```
