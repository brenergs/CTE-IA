```python
d = (40, 52, 55, 60, 70, 75, 86, 90, 90, 90, 94, 94, 95, 98, 100, 115, 125, 150)
```

## Utilizando Numpay

```python
import numpy as np
import matplotlib.pyplot as plt

data = [d]

print("Mínimo: ", np.min(data))
print("Máximo: ", np.max(data))
print("1º Quartil", np.quantile(data, 0.25))
print("Mediana: ", np.median(data))
print("3º Quartil", np.quantile(data, 0.75))

plt.boxplot(data)
plt.grid(True)
plt.show()
```

## Utilizando Pandas

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({'dados':d})

df.describe()
```

```python
plt.boxplot(df)
plt.grid(True)
plt.show()
```

## Identificação de outliers


```python
Q1 = df['dados'].quantile(0.25)
Q3 = df['dados'].quantile(0.75)

IQR = Q3 - Q1
Limite_Inferior = Q1 - 1.5 * IQR
Limite_Superior = Q3 + 1.5 * IQR

print("Limite Inferior: ", Limite_Inferior)
print("Limite Superior: ", Limite_Superior)
```

## Análise Final

* A distribuição dos dados não é perfeitamente simétrica, a cauda é mais longa para valores menores.
* A média (87.72) é ligeiramente menor que a mediana (90.0), e a distância da mediana ao 1º Quartil (90.0 - 71.25 = 18.75) é maior que a distância da mediana ao 3º Quartil (97.25 - 90.0 = 7.25).
* O valor 150 está a cima do limite superior (136.25), então, é um possível outlier.
