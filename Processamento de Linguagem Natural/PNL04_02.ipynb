## Questão 1: Tokenização com BERT

Crie um tokenizer e implemente uma função para tokenizar textos.

**Você deve criar:**

1. `tokenizer_bert`: Tokenizer carregado a partir de `MODELO_BASE`

2. `tokenizar_texto(texto)`: Função que recebe um texto e retorna dicionário com:
   - `'input_ids'`: Lista de IDs dos tokens
   - `'attention_mask'`: Máscara de atenção
   - `'tokens'`: Lista de tokens como strings
   - `'num_tokens'`: Número de tokens

**Use:** `max_length=128`, `truncation=True`, `padding=False`

```python
# === SEU CÓDIGO AQUI ===
tokenizer_bert = AutoTokenizer.from_pretrained(MODELO_BASE)

def tokenizar_texto(texto):
    inputs = tokenizer_bert(texto, max_length=128, truncation=True, padding=False)

    # Converte os IDs de volta para a lista de tokens em strings
    tokens = tokenizer_bert.convert_ids_to_tokens(inputs['input_ids'])

    return {
        'input_ids': inputs['input_ids'],
        'attention_mask': inputs['attention_mask'],
        'tokens': tokens,
        'num_tokens': len(tokens)
    }
```

## Questão 2: Extração de Embeddings

Carregue o modelo BERT e implemente uma função para extrair embeddings.

**Você deve criar:**

1. `modelo_bert`: Modelo carregado a partir de `MODELO_BASE`

2. `extrair_embedding_cls(texto)`: Função que extrai o embedding do token [CLS]
   - Retorna: Array numpy 1D com o embedding

**Dica:** Use `torch.no_grad()` para desabilitar gradientes

```python
# === SEU CÓDIGO AQUI ===
modelo_bert = AutoModel.from_pretrained(MODELO_BASE)

def extrair_embedding_cls(texto):
    inputs = tokenizer_bert(texto, return_tensors='pt', max_length=128, truncation=True, padding=False)

    with torch.no_grad():
        output = modelo_bert(**inputs)

    cls_embedding = output.last_hidden_state[0, 0, :].numpy()

    return cls_embedding
```

## Questão 3: Similaridade Semântica

Implemente função para calcular similaridade entre textos e crie uma matriz de similaridade.

**Você deve criar:**

1. `calcular_similaridade_textos(texto1, texto2)`: Função que calcula similaridade de cosseno entre dois textos
   - Retorna: float (similaridade de cosseno)

2. `matriz_similaridade`: Matriz numpy 4x4 com similaridades entre os textos de `TEXTOS_SIMILARIDADE`

**Dica:** Use sua função `extrair_embedding_cls` e `cosine_similarity` do sklearn

```python
# === SEU CÓDIGO AQUI ===
def calcular_similaridade_textos(texto1, texto2):
    # Extrai os embeddings (1D)
    emb1 = extrair_embedding_cls(texto1).reshape(1, -1)
    emb2 = extrair_embedding_cls(texto2).reshape(1, -1)

    # Calcula a similaridade de cosseno e retornamos como float
    sim = cosine_similarity(emb1, emb2)
    return float(sim[0][0])

# Gera a matriz 4x4 para os textos
n = len(TEXTOS_SIMILARIDADE)
matriz_similaridade = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        matriz_similaridade[i, j] = calcular_similaridade_textos(
            TEXTOS_SIMILARIDADE[i],
            TEXTOS_SIMILARIDADE[j]
        )
```

```python
# Testar similaridade
print("Similaridades entre pares de textos:")
print(f"T1-T2 (ambos positivos): {calcular_similaridade_textos(TEXTOS_SIMILARIDADE[0], TEXTOS_SIMILARIDADE[1]):.4f}")
print(f"T3-T4 (ambos negativos): {calcular_similaridade_textos(TEXTOS_SIMILARIDADE[2], TEXTOS_SIMILARIDADE[3]):.4f}")
print(f"T1-T3 (positivo-negativo): {calcular_similaridade_textos(TEXTOS_SIMILARIDADE[0], TEXTOS_SIMILARIDADE[2]):.4f}")

print(f"\nMatriz de Similaridade:")
print(matriz_similaridade.round(3))
```

## Questão 6: Classificador com Fine-tuning

Crie uma classe que encapsula o treinamento e predição de um classificador BERT.

**A classe `ClassificadorBERT` deve ter:**

1. `__init__()`: Inicializa tokenizer e modelo base

2. `treinar(textos, labels, epochs=2)`: Treina o modelo
   - Deve definir `self.treinado = False` na inicialização e `self.treinado = True` após treino

3. `prever(textos)`: Retorna lista de predições (0 ou 1)

4. `avaliar(textos, labels)`: Retorna dicionário com métricas
   - Deve incluir `'accuracy'`

**Use:** `MODELO_BASE` e configurações otimizadas para CPU

```python
class ClassificadorBERT:
    """
    Classificador de texto usando BERT com fine-tuning.
    """
    def __init__(self):
        self.tokenizer = AutoTokenizer.from_pretrained(MODELO_BASE)
        self.modelo = AutoModelForSequenceClassification.from_pretrained(MODELO_BASE, num_labels=2).to(device)
        self.treinado = False

    def treinar(self, textos, labels, epochs=2):
        # Preparar dados
        dataset = Dataset.from_dict({
            'text': textos,
            'label': labels
        })

        def tokenize_function(examples):
            return self.tokenizer(examples['text'], truncation=True, padding='max_length', max_length=128)

        tokenized_dataset = dataset.map(tokenize_function, batched=True)
        tokenized_dataset = tokenized_dataset.remove_columns(['text'])
        tokenized_dataset = tokenized_dataset.rename_column('label', 'labels')
        tokenized_dataset.set_format('torch')

        # Treinar modelo
        training_args = TrainingArguments(
            output_dir='./results',
            num_train_epochs=epochs,
            per_device_train_batch_size=4,
            weight_decay=0.01,
            logging_dir='./logs',
            logging_steps=5,
            eval_strategy='no',
            save_strategy='no',
            use_cpu=(device.type == 'cpu')
        )

        trainer = Trainer(
            model=self.modelo,
            args=training_args,
            train_dataset=tokenized_dataset,
            data_collator=DataCollatorWithPadding(tokenizer=self.tokenizer)
        )

        trainer.train()
        self.treinado = True

    def prever(self, textos):
        inputs = self.tokenizer(textos, padding=True, truncation=True, max_length=128, return_tensors='pt').to(device)
        self.modelo.eval()
        with torch.no_grad():
            outputs = self.modelo(**inputs)
        preds = torch.argmax(outputs.logits, dim=-1).cpu().tolist()
        return preds

    def avaliar(self, textos, labels):
        preds = self.prever(textos)
        acc = accuracy_score(labels, preds)
        return {'accuracy': acc}
```

```python
# Testar classificador
print("Testando ClassificadorBERT:")
print("=" * 50)

clf = ClassificadorBERT()
print("\nTreinando modelo (isso pode levar alguns minutos)...")

# Treinar
clf.treinar(
    DADOS_CLASSIFICACAO['treino']['textos'],
    DADOS_CLASSIFICACAO['treino']['labels'],
    epochs=2
)
print(f"Treinado: {clf.treinado}")

# Avaliar
metrics = clf.avaliar(
    DADOS_CLASSIFICACAO['teste']['textos'],
    DADOS_CLASSIFICACAO['teste']['labels']
)
print(f"\nAcurácia no teste: {metrics['accuracy']:.2%}")

# Predições em novos textos
novos = ["Produto fantástico!", "Muito ruim."]
preds = clf.prever(novos)
print(f"\nPredições:")
for texto, pred in zip(novos, preds):
    label = "POSITIVO" if pred == 1 else "NEGATIVO"
    print(f"  {texto} -> {label}")
```
